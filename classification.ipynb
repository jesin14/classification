```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score
```

# Loading and Preprocessing (2 marks):
● Load the breast cancer dataset from sklearn.
● Preprocess the data to handle any missing values and perform necessary feature
scaling.
● Explain the preprocessing steps you performed and justify why they are
necessary for this dataset.

Loading Dataset

```python
cancer = load_breast_cancer()
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df["target"] = cancer.target
df.head()
```

checking missing values


```python
df.isnull().sum().sum()
```

There is no missing values to handle.

Separating Features and Target

```python
X = df.drop("target",axis=1)
y = df["target"]
print("Feature shape: ",X.shape)
print("Target shape:", y.shape)
```

Handling Missing Values

```python

imputer = SimpleImputer(strategy="median")

X_imputed = pd.DataFrame(
    imputer.fit_transform(X),
    columns=X.columns
)

print("Missing values after preprocessing:", X_imputed.isnull().sum().sum())
```

I use median imputation to handle any missing value

```python

X_train, X_test, y_train, y_test = train_test_split(
    X_imputed,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

We divide the dataset into 80% training data and 20% testing data.

Feature Scaling

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed successfully!")
```

The features have different numerical ranges. StandardScaler transforms the features to have approximately zero mean and unit variance

## Classification Algorithm Implementation (5 marks):
Implement the following five classification algorithms:
1. Logistic Regression
2. Decision Tree Classifier
3. Random Forest Classifier
4. Support Vector Machine (SVM)
5. k-Nearest Neighbors (k-NN)
For each algorithm, provide a brief description of how it works and why it might be
suitable for this dataset.

Logistic Regression

```python
lo_model = LogisticRegression()
lo_model.fit(X_train_scaled,y_train)
logistic_predictions = lo_model.predict(X_test_scaled)
logistic_accuracy = accuracy_score(y_test, logistic_predictions)
print("Logistic Regression Accuracy:", logistic_accuracy)
```

Logistic Regression is a supervised classification algorithm that estimates the probability of a class. It is suitable for this dataset because it can classify tumors into two categories: malignant and benign.

Decision Tree Classifier

```python
tree_model = DecisionTreeClassifier(random_state=42)
tree_model.fit(X_train, y_train)
tree_predictions = tree_model.predict(X_test)
tree_accuracy = accuracy_score(y_test, tree_predictions)
print("Decision Tree Accuracy:", tree_accuracy)
```

A Decision Tree classifies data by splitting it into branches based on feature values. It is suitable because it can learn non-linear patterns and is relatively easy to understand.

Random Forest Classifier

```python

forest_model = RandomForestClassifier(n_estimators=100, random_state=42)
forest_model.fit(X_train, y_train)
forest_predictions = forest_model.predict(X_test)
forest_accuracy = accuracy_score(y_test, forest_predictions)
print("Random Forest Accuracy:", forest_accuracy)
```

Random Forest combines predictions from multiple decision trees to produce a final classification. It is suitable for this dataset because combining trees can improve predictive performance and reduce overfitting compared with a single tree.

SVM

```python
svm_model = SVC()
svm_model.fit(X_train_scaled, y_train)
svm_predictions = svm_model.predict(X_test_scaled)
svm_accuracy = accuracy_score(y_test, svm_predictions)
print("SVM Accuracy:", svm_accuracy)
```

SVM finds a decision boundary that separates classes while aiming to maximize the margin between them. 

kNN

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train_scaled, y_train)
knn_predictions = knn_model.predict(X_test_scaled)
knn_accuracy = accuracy_score(y_test, knn_predictions)
print("k-NN Accuracy:", knn_accuracy)
```

k-NN classifies a new observation based on the classes of its nearest neighbors. We use five neighbors in this model


## mODEL Comparison (2 marks):
● Compare the performance of the five classification algorithms.
● Which algorithm performed the best and which one performed the worst?

```python

results = pd.DataFrame({
    "Algorithm": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "SVM",
        "k-NN"
    ],
    "Accuracy": [
        logistic_accuracy,
        tree_accuracy,
        forest_accuracy,
        svm_accuracy,
        knn_accuracy
    ]
})


```

```python
results = results.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

results
```


We evaluated the five classification algorithms using accuracy on the testing data.

- Logistic Regression achieved an accuracy of 98.25%.
- Support Vector Machine (SVM) achieved an accuracy of 98.25%.
- Random Forest achieved an accuracy of 95.61%.
- k-Nearest Neighbors (k-NN) achieved an accuracy of 95.61%.
- Decision Tree achieved an accuracy of 91.23%.

The results show that Logistic Regression and SVM performed best, while Decision Tree performed worst in terms of accuracy.


```python

best_model = results.iloc[0]
worst_model = results.iloc[-1]

print("Best Performing Model:", best_model["Algorithm"])
print("Best Accuracy:", round(best_model["Accuracy"] * 100, 2), "%")

print("\nWorst Performing Model:", worst_model["Algorithm"])
print("Worst Accuracy:", round(worst_model["Accuracy"] * 100, 2), "%")
```
